## Export labels (COCO $\to$ YOLO)

https://docs.ultralytics.com/guides/coco-to-yolo#quick-start

```python
df_train_meta["image_path"].iloc[0].split("/")[-1]
```

```python
labels_dir = Path(
    "data/f5ddba02-4295-48d9-8ad5-ba8583871323/"
    "bc6ba36c-aa02-4630-a0ea-4150b9b69e36/"
)

save_dir = labels_dir / "yolo_converted"
labels_dir.mkdir(parents=True, exist_ok=True)

convert_coco(
    labels_dir=str(labels_dir),
    save_dir=str(save_dir),
    cls91to80=False,
    use_segments=True
)
```

## View bounding boxes

```python
for i in range(10):
    image_path = "data/"+df_train_meta.loc[i, "image_path"]
    image = PIL_Image.open(image_path).convert("RGB")
    annotated = image.copy()
    draw = ImageDraw.Draw(annotated)
    for bbox in df_train_meta['labels'].iloc[i]:
        draw.rectangle(bbox, outline = "red", width = 3)
    display(annotated)
```

```python
variables = ['visibility_original', 'region_original', 
             'image_angle_original', 'pavement_color_original', 'environment_original', 
             'weather_condition_original', 'truck_color_original']
df_train_vars = df_train_meta[['image_path', 'labels']+variables]
df_train_vars
```

## By level

```python
for var in variables:
    levels = list(df_train_vars[var].unique())
    for level in levels:
        print(f"\n------ {var} = {level} ------\n")
        imgs = df_train_vars[df_train_vars[var] == level]["image_path"][:5]
        #print(imgs)
        for img in imgs:
            print(img)
            display(Image(filename="data/"+img))
```
